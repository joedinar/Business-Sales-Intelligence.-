# Exploratory Data Analysis

```python
Supstore['Order Month'] = Supstore['Order Date'].dt.month_name()
```

```python
Supstore[['Order Date', 'Order Month']].head()
```

```python
# Create month number and month name
Supstore['Order Month Number'] = Supstore['Order Date'].dt.month
Supstore['Order Month'] = Supstore['Order Date'].dt.month_name()

# Calculate monthly sales by category
monthly_sales = Supstore.groupby(
    ['Order Month Number', 'Order Month', 'Category']
)['Sales'].sum().reset_index()

# Sort months correctly
monthly_sales = monthly_sales.sort_values('Order Month Number')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for category in monthly_sales['Category'].unique():
    
    data = monthly_sales[
        monthly_sales['Category'] == category
    ]
    
    plt.plot(
        data['Order Month Number'],
        data['Sales'],
        marker='o',
        label=category
    )

# Show month names on x-axis
months = ['January', 'February', 'March', 'April', 'May', 'June',
          'July', 'August', 'September', 'October', 'November', 'December']

plt.xticks(range(1, 13), months, rotation=45)

plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.title('Monthly Sales by Product Category')
plt.legend(title='Category')

plt.savefig(r'D:\SQL\Project Simulation\line_cat.png', dpi=300, bbox_inches='tight')

plt.tight_layout()
plt.show()
```

```python
# Create month number and month name

Supstore['Order Year'] = Supstore['Order Date'].dt.year

# Calculate monthly sales by category
yearly_sales = Supstore.groupby(
    ['Order Year', 'Category']
)['Sales'].sum().reset_index()
```

```python
plt.figure(figsize=(10, 6))

for category in yearly_sales['Category'].unique():
    
    data = yearly_sales[
        yearly_sales['Category'] == category
    ].sort_values('Order Year')
    
    plt.plot(
        data['Order Year'],
        data['Sales'],
        marker='o',
        label=category
    )

plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.title('Yearly Sales by Product Category')

plt.xticks(sorted(yearly_sales['Order Year'].unique()))
plt.legend(title='Category')

plt.savefig(r'D:\SQL\Project Simulation\yearly_sales.png', dpi=300, bbox_inches='tight')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Make sure the column is datetime type
Supstore['Order Date'] = pd.to_datetime(Supstore['Order Date'])

# Extract the year into a new column
Supstore['Order Year'] = Supstore['Order Date'].dt.year
```

```python
Supstore.head()
```

```python
# Aggregate sales by year
yearly_sales = Supstore.groupby('Order Year')['Sales'].sum().reset_index()

# Rename columns first
yearly_sales.columns = ['year', 'total_sales']

# Now compute YoY growth
yearly_sales['yoy_growth_%'] = yearly_sales['total_sales'].pct_change() * 100

yearly_sales
```

```python
plt.figure(figsize=(8, 5))
plt.plot(yearly_sales['year'], yearly_sales['yoy_growth_%'], marker='o', linewidth=2)

plt.title('YoY Sales Growth (%)')
plt.xlabel('Year')
plt.ylabel('YoY Growth (%)')
plt.axhline(0, color='gray', linestyle='--', linewidth=1)  # reference line at 0%
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(yearly_sales['year'], yearly_sales['yoy_growth_%'], marker='o', linewidth=2)

plt.title('YoY Sales Growth (%)')
plt.xlabel('Year')
plt.ylabel('YoY Growth (%)')
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.grid(True, alpha=0.3)

# Force x-axis to show whole numbers only
plt.gca().xaxis.set_major_locator(mtick.MaxNLocator(integer=True))

plt.tight_layout()
plt.show()
```
